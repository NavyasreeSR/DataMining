```python
import pandas as pd
from datetime import datetime, timedelta


# Convert 'Date' column to datetime
data['Date'] = pd.to_datetime(data['Date'])

# Calculate average time difference between consecutive purchases for each item
data['time_diff'] = data.groupby('Item')['Date'].diff().dt.days

# Calculate average time difference
average_time_diff = data.groupby('Item')['time_diff'].mean().to_dict()

# Predict next purchase date for each item
next_purchase_dates = {}
for item, avg_diff in average_time_diff.items():
    last_purchase_date = data[data['Item'] == item]['Date'].max()
    if pd.notna(last_purchase_date):
        next_purchase_dates[item] = last_purchase_date + timedelta(days=avg_diff)
    else:
        # If there's no previous purchase date, predict next purchase date as today
        next_purchase_dates[item] = datetime.now()

# Recommend items to be purchased next within the year
recommendations = []
current_date = datetime.now()
for item, next_purchase_date in next_purchase_dates.items():
    if next_purchase_date.year == current_date.year:
        recommendations.append((item, next_purchase_date))

# Sort recommendations by next purchase date
recommendations.sort(key=lambda x: x[1])

# Print recommendations
print("Items to be purchased next within the year:")
for i, (item, next_purchase_date) in enumerate(recommendations, 1):
    print(f"{i}. Item: {item}, Next Purchase Date: {next_purchase_date.strftime('%Y-%m-%d')}")
```

```python
import pandas as pd
from datetime import datetime, timedelta

# Convert 'Date' column to datetime
data['Date'] = pd.to_datetime(data['Date'])

# Calculate average time difference between consecutive purchases for each item
data['time_diff'] = data.groupby('Item')['Date'].diff().dt.days

# Calculate average time difference
average_time_diff = data.groupby('Item')['time_diff'].mean().to_dict()

# Calculate average quantity purchased for each item and round to nearest integer
average_quantity = data.groupby('Item')['Quantity'].mean().round().astype(int).to_dict()

# Predict next purchase date and quantity for each item
next_purchase_info = {}
for item, avg_diff in average_time_diff.items():
    last_purchase_date = data[data['Item'] == item]['Date'].max()
    if pd.notna(last_purchase_date):
        next_purchase_date = last_purchase_date + timedelta(days=avg_diff)
    else:
        # If there's no previous purchase date, predict next purchase date as today
        next_purchase_date = datetime.now()

    # Predict quantity
    predicted_quantity = average_quantity.get(item, 0)  # Default to 0 if no previous data is available

    next_purchase_info[item] = {'Next Purchase Date': next_purchase_date, 'Predicted Quantity': predicted_quantity}

# Recommend items to be purchased next within the year
recommendations = []
current_date = datetime.now()
for item, info in next_purchase_info.items():
    next_purchase_date = info['Next Purchase Date']
    if next_purchase_date.year == current_date.year:
        recommendations.append((item, next_purchase_date, info['Predicted Quantity']))

# Sort recommendations by next purchase date
recommendations.sort(key=lambda x: x[1])

# Print recommendations
print("Items to be purchased next within the year:")
for i, (item, next_purchase_date, predicted_quantity) in enumerate(recommendations, 1):
    print(f"{i}. Item: {item}, Next Purchase Date: {next_purchase_date.strftime('%Y-%m-%d')}, Predicted Quantity: {predicted_quantity}")

# Create a DataFrame from recommendations
df_recommendations = pd.DataFrame(recommendations, columns=['Item', 'Next Purchase Date', 'Predicted Quantity'])
```
